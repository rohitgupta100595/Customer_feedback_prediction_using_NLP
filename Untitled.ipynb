## **Imports and Configuration**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import string
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import CountVectorizer,TfidfTransformer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report,confusion_matrix
import joblib
```

## **Data Loading**

```python
df = pd.read_csv('sentiment-analysis.csv',sep = ', ',engine='python')
```

## **Data Quality and Validation**

### Missing Values

```python
df.isnull().sum()
```

```python
# Dropping na
df.dropna(inplace=True)
```

### Formating Texts and column names

```python
df['"Text'] = df['"Text'].str.replace('"','')
df['Confidence Score"'] = df['Confidence Score"'].str.replace('"','').astype(float)
df.rename(columns={'"Text':'Text','Confidence Score"':'Confidence Score'},inplace=True)
```

### Duplicate Values

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

### Checking Unique values for each variable

```python
def unique_vals(df):
    result = pd.DataFrame({
        'Unique Counts':df.nunique(),
        'Unique Vals':[df[col].unique().tolist() for col in df.columns],
        'Data Type':df.dtypes
    })
    return result
```

```python
unique_vals(df)
```

### Extracting text length

```python
df['Length'] = df['Text'].apply(len)
```

### Remove Unwanted Columns

```python
df.drop(['Source','User ID'],axis=1,inplace=True)
```

## **Text-Processing**

### Text-Processing Fuction to remove punctuation and stopwords

```python
def text_proc(text):
    nopunc = [char for char in text if char not in string.punctuation]
    nopunc = ''.join(nopunc)
    return [word for word in nopunc.split() if word.lower() not in stopwords.words('english')]
```

```python
df['clean_text'] = df['Text'].apply(text_proc)
```

## **Train-Test Split**

```python
X = df['Text']
y = df['Sentiment']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=101)
```

## **Baseline Model**

### Creating Pipeline for the process flow

```python
pipeline = Pipeline([
    ('bow',CountVectorizer(analyzer=text_proc)),
    ('tfidf',TfidfTransformer()),
    ('classifier',MultinomialNB())
])
```

## **Model Training**

```python
pipeline.fit(X_train,y_train)
```

## **Model Evaluation**

```python
y_pred = pipeline.predict(X_test)
```

```python
print(classification_report(y_test,y_pred))
```

```python
print(confusion_matrix(y_test,y_pred))
```

## **Model Comaprison**

```python
from sklearn.linear_model import LogisticRegression
```

```python
pipeline_m1 = Pipeline([
    ('bow',CountVectorizer(analyzer=text_proc)),
    ('tfidf',TfidfTransformer()),
    ('classifier',LogisticRegression())
])
```

```python
pipeline_m1.fit(X_train,y_train)
```

```python
y_pred = pipeline_m1.predict(X_test)
```

```python
print(classification_report(y_test,y_pred))
```

```python
print(confusion_matrix(y_test,y_pred))
```

From the above two models (**Naive-bayes** and **Logistic Regression**), Naive-bayes performed well than the Logistice Regression. hence we will choose **Naive-bayes** for our classifier.

## **Final Pipeline/Model Output**

```python
joblib.dump(pipeline,'sentiment_pipeline.pkl')
```
